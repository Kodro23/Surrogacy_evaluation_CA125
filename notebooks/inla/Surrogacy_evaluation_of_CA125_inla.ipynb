## Load and clean dataset

```r
#Load dataset
ca125 <- read.csv("./data/MAOV_ca125_all.csv", sep=";")
ca125 <- ca125%>%mutate(id=paste(trialid,"_",patid)) 
ca125 %>% summarize(
    n_patients=n_distinct(id),
    n_trials=n_distinct(trialid),
    n_measures=n()
)
```

```r
#Clean dataset
#check time between first surgery and first measure
Sys.setlocale("LC_TIME", "C")
ca125$delay_surg_trt<-(as.Date(toupper(ca125$dotrt_begin), format = "%d%b%Y")-as.Date(toupper(ca125$dosurg1), format = "%d%b%Y"))
#delete patients with delay <7
ca125<- ca125[!ca125$id %in% unique(ca125$id[ca125$delay_surg_trt < 7 & !is.na(ca125$delay_surg_trt)]),]
```

```r
# Clean first progression column
ca125 <- ca125 %>%
  mutate(
    dor = as.Date(toupper(dor), format = "%d%b%Y"),
    dofirst_prog = as.Date(toupper(dofirst_prog), format = "%d%b%Y"),
    dofirst_prog = if_else(
      is.na(dofirst_prog) & !is.na(dor),
      dor + pfs_time_m * 30.4375,
      dofirst_prog
    )
  )
ca125 <- ca125 %>%
  mutate(
    dotrt_begin = as.Date(toupper(dotrt_begin), format = "%d%b%Y"),
    donext_trt = as.Date(toupper(donext_trt), format = "%d%b%Y"),
    do_CA125 = as.Date(toupper(do_CA125), format = "%d%b%Y")
  )
```

```r
#delete measures after progression 
ca125<-ca125 %>%
  filter( (trialid != "MRC-ICON7" & do_CA125 <= dofirst_prog) |
        (trialid == "MRC-ICON7" & time_ca125<=pfs_time_m * 30.4375))
```

```r
ca125 %>% summarize(
    n_patients=n_distinct(id),
    n_trials=n_distinct(trialid),
    n_measures=n()
)
```

```r
#Create variables
ca125$trt_bin<- if_else(ca125$trt=="Standard regimen", 0,1)
ca125["os_time_d"]<-ca125$os_time_m*30.4375
ca125["pfs_time_d"]<-ca125$pfs_time_m*30.4375
ca125["os_statut"]<-factor(ifelse(ca125$os_status=='Yes', 1, 0), levels=c(0,1))
ca125["pfs_statut"]<-factor(ifelse(ca125$PFS_status=='Yes', 1, 0), levels=c(0,1))
#restrict data                             
colonnes<-c("trialid","patid","id",'trt','dor','CA125_mes','do_CA125','id_temp','dotrt_begin', "donext_trt",'dotrt_end',"time_ca125",'os_status','OS_time','PFS_status','PFS_time',"nb_mesures",
"ca125_base",'Visit','pfs_time_m',"pfs_time_d",'os_time_m',"os_time_d",'group','country','maov_group','dob','age','figo','figo4c',"trt_bin","os_statut","pfs_statut")
ca125_all_6 <- ca125[colonnes]%>%
  filter(time_ca125 <= 188) %>%
  group_by(id) %>%
  filter(n() > 1) %>%
  ungroup()
ca125_all_6["id_num"]<- as.numeric(as.factor(ca125_all_6$id))
ca125_all_6["trialid_num"]<- as.numeric(as.factor(ca125_all_6$trialid))
ca125_all_6["log_ca125"]<-log(ca125_all_6$CA125_mes)
```

# Joint modeling

## Specification

**Longitudinal model**
$
Y_i(t) = (\beta_0 +b_{0i})+ \sum_{k=1}^{2} (\beta_k+b_{ki}) \, ns_k(t)+ \sum_{k=1}^{2} \gamma_k \, ns_k(t)*T_i+ \varepsilon_i(t)
$

$
b_i \sim \mathcal{N}(0, D), \quad \varepsilon_i(t) \sim \mathcal{N}(0, \sigma^2)
$\
**Survival model**
$
h_i(t) = h_0(t)\exp(\alpha \, T_i)
$

$
h_0(t) = \sum_{m=1}^{M} \lambda_m B_m(t)
$

**Joint association structure**

$
h_i(t \mid b_i) = h_0(t)\exp(\alpha \, T_i + \eta^\top b_i)
$\
**Full likelihood**

$
L = \prod_{i=1}^N 
\int 
f(Y_i \mid b_i)\,
f(T_i, \delta_i \mid b_i)\,
f(b_i)\, db_i
$

**Longitudinal model details**

With splines :
- s trials
- $n_i$ patients in trial $i$, $n = \sum_{i=1}^{s} n_i$ the total number of patients
- $m_{ij}$ the number of measures for patient j in study i, $m_{i}=\sum_{j=1}^{n_i} m_{ij}$ the total number of observations in trial $i$, $N=\sum_{i=1}^{s} m_i=\sum_{i=1}^{s}\sum_{j=1}^{n_i} m_{ij}$ the total number of observations
- $y(t_{ijk})$ the $k^{th}$ value $k \in {1,...,m_{ij}}$ of biomarker CA-125 for patient $j$, $j \in {1,...,n_i}$, in study $i$, $i \in {1,...,s}$
- 
$$
T_{ij}=
\begin{cases} 
1 & \text{if patient $j$ in trial $i$ received the treatment} 
\\ 0 & \text{otherwise}
\end{cases}
$$

Mixed-effects model for varataion of $y$:
$
log[y(t_{ijk})]= f(t_{ijk}) +\epsilon_{ijk}
$, 
with:
- $ f(t)= (\theta_{1}+b_{1i}+b_{1ij}) + (\theta_{3}+b_{3ij})t+ (\theta_{2}+ b_{2,ij})B_{2}(t)+ (\theta_{4}+ b_{4,i})T_{ij}*t+ (\theta_{5}+ b_{5,i}) T_{ij}*B_{2}(t)$; $B_{2}(t) = d_{0} - d_2; d_k=\frac{(t-q_k(t))_{+}^3 - (t-q_2(t))_{+}^3}{q_k(t) - q_2(t)}$
- $x_+ = max (0, x)  $,
- $q_0(t), q_1(t), q_2(t)  \text{ are quantiles of t at 18\%, 50\%, and 98\%} $ (7, 64, 180 days)
- $(b_{1i}, b_{4i},b_{5i})^\top \sim N(0, \phi)$, $\phi = \begin{pmatrix} \sigma_{1}^2 & \sigma_1\sigma_4 &\sigma_1\sigma_5\\ \sigma_1\sigma_4 & \sigma_{4}^2 & \sigma_4\sigma_5 \\ \sigma_1\sigma_5 & \sigma_4\sigma_5 & \sigma_{5}^2\end{pmatrix}$  (random effect at study level)
- $ (b_{1ij},b_{3ij},b_{2ij})^t \sim N(0, D)$, $D = \begin{pmatrix} \sigma_{b1}^2 & \sigma_{b1b3} & \sigma_{b1b2} \\ \sigma_{b1b3} & \sigma_{b3}^2 &  \sigma_{b2b3} \\ \sigma_{b1b2} & \sigma_{b2b3} &\sigma_{b2}^2 \end{pmatrix}$(random effects at patient-level)
- $\epsilon_{ijk} \sim N(0, \sigma^2_\epsilon I_N)$


**Conditional BLUP for patient-level random effects**:

- $\theta=(\theta_1 \text{  } \theta_3 \text{  } \theta_2 \text{  } \theta_4 \text{  } \theta_5)^T$

Matricial form: $ y= X\theta + Z_sb_s + Z_{n}b_{n} + \epsilon$, with :
$
y =
\begin{pmatrix}
y(t_{111}) \\
y(t_{112}) \\
\vdots \\
y(t_{11 m_{11}}) \\
y(t_{121}) \\
\vdots \\
y(t_{1 n_1 m_{1m=n_1}}) \\
\vdots \\
y(t_{s n_s m_{sn_s}})
\end{pmatrix}
\in \mathbb{R}^{N},
\qquad
$,
$
X =
\begin{pmatrix}
1 & t_{111} & B_2(t_{111}) & T_{11}*t_{111} & T_{11}*B_2(t_{111}) \\
1 & t_{112} & B_2(t_{112}) & T_{11}*t_{112} & T_{11}*B_2(t_{112})\\
\vdots & \vdots & \vdots & \vdots & \vdots\\
1 & t_{11 m_{11}} & B_2(t_{11 m_{11}}) & T_{11}*t_{11 m_{11}} & T_{11}*B_2(t_{11 m_{11}}) \\
1 & t_{121} & B_2(t_{121}) & T_{12}*t_{121} & T_{12}*B_2(t_{121})\\
\vdots & \vdots & \vdots & \vdots & \vdots\\
1 & t_{s n_s m_{sn_s}} & B_2(t_{s n_s m_{sn_s}}) & T_{sn_s}*t_{t_{s n_s m_{sn_s}}} & T_{sn_s}*B_2(t_{t_{s n_s m_{sn_s}}}) 
\end{pmatrix}
\in \mathbb{R}^{N \times 5}
$
- $b_s \sim N(0, \Phi)$, $\Phi = \mathrm{diag}(\phi,...,\phi) \in \mathbb{R}^{3s \times 3s}$ (study-level random intercept)
$$
b_s =
\begin{pmatrix}
b_{11} & b_{41} &  b_{51} \\
b_{12} & b_{42} &  b_{52} \\
\vdots & \vdots & \vdots \\
b_{1s} & b_{4s} & b_{5s} 
\end{pmatrix}
\in \mathbb{R}^{s \times 3}
$$
- $b_{n} \sim N(0, \Delta)$ (patient-level random intercept+slope), $
\Delta = \mathrm{diag}(D,...,D) \in \mathbb{R}^{3n \times 3n}
$

$$
b_{n} =
\begin{pmatrix}
b_{1,11} \\
b_{3,11} \\
b_{2,11} \\
b_{1,12} \\
b_{3,12} \\
b_{2,12} \\
\vdots \\
b_{1,s n_s} \\
b_{3,s n_s} \\
b_{2,s n_s}
\end{pmatrix}
\in \mathbb{R}^{3n},
\qquad
$$

- $\epsilon \sim N(0, \sigma^2_{\epsilon}I_N)$

$\boxed{Var(y)=V=Z_s \Phi Z_s^T + Z_{n}\Delta Z_{n}^T + \sigma^2_{\epsilon}I_N}$\
with:
- $ Z_s =\begin{pmatrix}\mathbf{Z}_{s_1} & 0 & \cdots & 0 \\0 & \mathbf{Z}_{s_2} & \cdots & 0 \\\vdots & \vdots & \ddots & \vdots \\0 & 0 & \cdots & \mathbf{Z}_{s_s}\end{pmatrix} \in \mathbb{R}^{N \times 3s}$
. For trial $i$, $ Z_{s_i} =\begin{pmatrix}1 & T_{i1}*t_{i11} & T_{i1}*B_2(t_{i11})\\1 & T_{i1}*t_{i12} & T_{i1}*B_2(t_{i12})\\\vdots & \vdots & \vdots\\1 & T_{in_i}*t_{in_im_i} & T_{in_i}*B_2(t_{in_im_i}) \end{pmatrix}\in \mathbb{R}^{m_{i} \times 3}$

For patient $(i,j)$,:

$
Z_{ij}=\begin{pmatrix}1 & t_{ij1} & B_2(t_{ij1})\\1 & t_{ij2}&  B_2(t_{ij2})\\\vdots & \vdots & \vdots\\1 & t_{ijm_{ij}}& B_2(t_{ijm_{ij}})\end{pmatrix}\in \mathbb{R}^{m_{ij} \times 3}$, 
$
Z_{n}=\begin{pmatrix}Z_{11} & 0 & \cdots & 0\\0 & Z_{12} & \cdots & 0\\\vdots & \vdots & \ddots & \vdots\\0 & 0 & \cdots & Z_{s n_s}\end{pmatrix} \in \mathbb{R}^{N \times 3n}
$

$$
\widehat{b}_{n} = \Delta Z_{n}^\top V^{-1}(y-X\hat{\theta})
$$
$
\begin{pmatrix}
\hat b_{1ij} \\
\hat b_{3ij} \\
\hat b_{2ij}
\end{pmatrix}
=[\Delta]_{ij}\sum_{k=1}^{m_{ij}}\begin{pmatrix}1\\ t_{ijk} \\B_2(t_{ijk})\end{pmatrix}\,\bigl[V^{-1}(y - X\hat\theta)\bigr]_{ijk}
$


**BLUP for Study-Level Random Effect**\
$$
\widehat{b}_{s} = \Phi Z_s^\top V^{-1} (y-X\hat{\theta}) 
$$
$
\begin{pmatrix}
\hat b_{1i} \\
\hat b_{4i} \\
\hat b_{5i}
\end{pmatrix}
=[\Phi]_{i} \sum_{j=1}^{n_i} \sum_{k=1}^{m_{ij}}\begin{pmatrix}1 \\T_{ij}*t_{ijk} \\T_{ij}*B_2(t_{ijk})\end{pmatrix}\,\bigl[V^{-1}(y - X\hat\theta)\bigr]_{ijk}
$



**Fixed Effects Estimate**\
Stack all data across studies:
$$
\hat{\beta} = (\mathbf{X}^\top \mathbf{V}^{-1} \mathbf{X})^{-1} \mathbf{X}^\top \mathbf{V}^{-1} \mathbf{y}
$$

## Surrogacy evaluation

**Compute $R^2_{trial}$**
- From joint model extract log(HR) associated with "trt_bin"
- Predict average log(CA-125) trajectory for treatment group and control group using joint model\
1) **Nadir:** $\Delta = min_t(\hat{Y}_{trt}) - min_t(\hat{Y}_{ctrl})$, $log(HR) = \alpha + \beta \Delta$ 
2) **Time to nadir:** $\Delta_t = \text{time to }min_t(\hat{Y}_{trt}) - \text{time to }min_t(\hat{Y}_{ctrl})$, $log(HR) = \alpha + \beta \Delta_t$
3) **Difference of slopes:** $S_i=\frac{\hat{Y}(t_i) -\hat{Y}(t_i-1)}{t_i - t_i-1}$, $i \in \{30,60,90,120,150,180\}$, $log(HR) = \alpha + \beta (S_{i_{trt}}- S_{i_{ctrl}})$
4) **Area between curves:** $AUC\approx \sum_{i=1}^{n-1} (t_{i+1} - t_i)\,\frac{(\hat{Y_{trt}}(t_i) -\hat{Y_{ctrl}}(t_i) )+ (\hat{Y}_{trt}(t_{i+1}) -\hat{Y}_{ctrl}(t_{i+1}))}{2}$, $log (HR)= \alpha + \beta AUC$
5) **Reduction from baseline:** $R=\frac{\hat{Y}(t_0) - \hat{y}(t_f)}{\hat{Y}(t_0)}$, $\Delta R = R_{trt} - R_{ctrl}$, $log (HR)= \alpha + \beta \Delta R$ (refaire à tous les mois)
6) **Absolute difference** $log(HR)= \alpha +\beta D_i$,  $D_i=\hat{Y}(t_i)_{trt} - \hat{Y}(t_i)_{ctrl}$, $t_i \in \{60,180\}$
7) **KELIM at 3 months**: The joint cumulative distribution of $PFS_{ij}$ and $KELIM_{ij}$ given $T_{ij}=trt$ is modeled with a one-parameter Placket copula $C_\theta$:
$F_{PFS_{ij},KELIM_{ij}} = C_{\theta} [ F_{PFS_{ij}}(t,trt), F_{KELIM_{ij}}(t,trt), \theta ] $, with $F_{PFS_{ij}}$ and $F_{KELIM_{ij}}$ the marginal cumulative distribution functions of PFS and KELIM respectively.  
$$
C_{\theta}(u,v)=
\begin{cases} 
\frac{1 + (u+v)(\theta-1)-\sqrt{[1 + (\theta-1)(u+v)]^2 + 4\theta(1-\theta)uv}}{2(\theta-1)} & \text{if $\theta \neq 1,$} 
\\ uv & \text{otherwise.}
\end{cases}
$$\\
with association parameter:
$\theta=\frac{\mathbb{P}(PFS_{ij} >t, KELIM_{ij}>k)\mathbb{P}(PFS_{ij} \leq t, KELIM_{ij} \leq k)}{ \mathbb{P}(PFS_{ij} >t, KELIM_{ij} \leq k)\mathbb{P}(PFS_{ij} \leq t, KELIM_{ij}>k)}$\\

**Significance of treatment effect on CA125**\
Treatment interraction with both spliens
$$
\boldsymbol{\gamma}
=
\begin{pmatrix}
\theta_4\\
\theta_5
\end{pmatrix},
$$
Null hypothesis: $H_0:\boldsymbol{\gamma}=\mathbf{0}.$

Posterior mean from samples: $\hat{\boldsymbol{\gamma}}=
\begin{pmatrix}
E(\gamma_1\mid y)\\
E(\gamma_2\mid y)
\end{pmatrix}
$. 

Covariance matrix

$$
\widehat{\Sigma}_{\gamma}
=
\begin{pmatrix}
\operatorname{Var}(\gamma_1\mid y)
&
\operatorname{Cov}(\gamma_1,\gamma_2\mid y)
\\
\operatorname{Cov}(\gamma_1,\gamma_2\mid y)
&
\operatorname{Var}(\gamma_2\mid y)
\end{pmatrix}.
$$

Wald statistic: $W=\hat{\boldsymbol{\gamma}}^{T}\widehat{\Sigma}_{\gamma}^{-1}\hat{\boldsymbol{\gamma}} \sim \chi^2_2 \text{ under } H_0 $

P-value: $p=P\left(\chi^2_2 \geq W\right).$


```r
# Significance of treatment effect on CA125 trajectory

# Kenward-roger wald test
gamma_draws <- cbind(theta4 = theta4,theta5 = theta5)
# Posterior mean vector
gamma_hat <- colMeans(gamma_draws)
# Joint posterior covariance matrix
V_gamma <- cov(gamma_draws)
# Wald statistic
W <- as.numeric(t(gamma_hat) %*%solve(V_gamma) %*%gamma_hat)
df <- length(gamma_hat)
p_value <- pchisq(W,df = df,lower.tail = FALSE)
data.frame(Wald_chisq = W,df = df,p_value = p_value)
```

```r
#Compute R^2 for summaries

surrogate_names <- c(
  delta_nadir      = "Diff. nadir",
  ratio_nadir      = "Ratio nadir",
  delta_time_nadir = "Time to nadir",
  delta_auc        = "AUC",

  delta_pct_m1 = "Reduction from baseline at 1 month",
  delta_pct_m2 = "Reduction from baseline at 2 months",
  delta_pct_m3 = "Reduction from baseline at 3 months",
  delta_pct_m4 = "Reduction from baseline at 4 months",
  delta_pct_m5 = "Reduction from baseline at 5 months",
  delta_pct_m6 = "Reduction from baseline at 6 months",

  ratio_pct_m1 = "Ratio from baseline at 1 month",
  ratio_pct_m2 = "Ratio from baseline at 2 months",
  ratio_pct_m3 = "Ratio from baseline at 3 months",
  ratio_pct_m4 = "Ratio from baseline at 4 months",
  ratio_pct_m5 = "Ratio from baseline at 5 months",
  ratio_pct_m6 = "Ratio from baseline at 6 months",

  delta_slope_m1 = "Slope at 1 month",
  delta_slope_m2 = "Slope at 2 months",
  delta_slope_m3 = "Slope at 3 months",
  delta_slope_m4 = "Slope at 4 months",
  delta_slope_m5 = "Slope at 5 months",
  delta_slope_m6 = "Slope at 6 months",

  ratio_slope_m1 = "Ratio slope at 1 month",
  ratio_slope_m2 = "Ratio slope at 2 months",
  ratio_slope_m3 = "Ratio slope at 3 months",
  ratio_slope_m4 = "Ratio slope at 4 months",
  ratio_slope_m5 = "Ratio slope at 5 months",
  ratio_slope_m6 = "Ratio slope at 6 months",

  absolutediff_m1 = "Absolute diff. at 1 month",
  absolutediff_m2 = "Absolute diff. at 2 months",
  absolutediff_m3 = "Absolute diff. at 3 months",
  absolutediff_m4 = "Absolute diff. at 4 months",
  absolutediff_m5 = "Absolute diff. at 5 months",
  absolutediff_m6 = "Absolute diff. at 6 months",
  
  relativediff_m1 = "Relative diff. at 1 month",
  relativediff_m2 = "Relative diff. at 2 months",
  relativediff_m3 = "Relative diff. at 3 months",
  relativediff_m4 = "Relative diff. at 4 months",
  relativediff_m5 = "Relative diff. at 5 months",
  relativediff_m6 = "Relative diff. at 6 months"
)

#Compute one R2 per posterior sample and surrogate

R2_posterior <- lapply(names(surrogate_names), function(var) {

  trial_samples %>%
    group_by(sample) %>%
    summarise(
      R2 = summary(
        lm(
          reformulate(var, response = "logHR"),
          data = pick(everything())
        )
      )$r.squared,
      .groups = "drop"
    ) %>%
    mutate(
      surrogate = var,
      Model = unname(surrogate_names[var])
    )

}) %>%
  bind_rows()

#Posterior summary of R2

R2_summary <- R2_posterior %>%
  group_by(surrogate, Model) %>%
  summarise(
    mean_R2 = mean(R2, na.rm = TRUE),
    lower_95 = quantile(R2, 0.025, na.rm = TRUE),
    upper_95 = quantile(R2, 0.975, na.rm = TRUE),
    .groups = "drop"
  ) %>%
  mutate(
    `R^2 [95% CI]` = sprintf(
      "%.3f [%.3f, %.3f]",
      mean_R2,
      lower_95,
      upper_95
    )
  ) %>%
  select(Model, `R^2 [95% CI]`)


#Restore desired ordering
R2_summary <- R2_summary %>%
  mutate(
    Model = factor(
      Model,
      levels = unname(surrogate_names)
    )
  ) %>%
  arrange(Model) %>%
  mutate(Model = as.character(Model))


R2_summary
```

```r
#Leave-one-trial-out R2

loso_surrogates <- c(
  delta_pct_m3 = "Reduction from baseline at 3 months",
  delta_pct_m6 = "Reduction from baseline at 6 months",
  ratio_pct_m3 = "Ratio from baseline at 3 months",
  ratio_pct_m6 = "Ratio from baseline at 6 months",

  delta_slope_m3 = "Slope at 3 months",
  delta_slope_m6 = "Slope at 6 months",
  ratio_slope_m3 = "Ratio slope at 3 months",
  ratio_slope_m6 = "Ratio slope at 6 months",

  absolutediff_m3 = "Absolute diff. at 3 months",
  absolutediff_m6 = "Absolute diff. at 6 months",
  relativediff_m3 = "Relative diff. at 3 months",
  relativediff_m6 = "Relative diff. at 6 months"
)

trials <- sort(unique(trial_samples$trialid_num))


loso_r2_posterior <- lapply(trials, function(left_out) {

  dat_loso <- trial_samples %>%
    filter(trialid_num != left_out)

  lapply(names(loso_surrogates), function(var) {

    dat_loso %>%
      group_by(sample) %>%
      summarise(
        R2 = summary(
          lm(
            reformulate(var, response = "logHR"),
            data = pick(everything())
          )
        )$r.squared,
        .groups = "drop"
      ) %>%
      mutate(
        left_out_trial = left_out,
        surrogate = var,
        summary = unname(loso_surrogates[var])
      )

  }) %>%
    bind_rows()

}) %>%
  bind_rows()

loso_boot_r2 <- loso_r2_posterior %>%
  group_by(left_out_trial, surrogate, summary) %>%
  summarise(
    mean_R2 = mean(R2, na.rm = TRUE),
    lower_95 = quantile(R2, 0.025, na.rm = TRUE),
    upper_95 = quantile(R2, 0.975, na.rm = TRUE),
    .groups = "drop"
  )
loso_boot_r2 <- loso_boot_r2 %>%
  left_join(
    trial_map,
    by = c("left_out_trial" = "trialid_num")
  ) %>%
  mutate(
    left_out_trial = trialid
  )
```
